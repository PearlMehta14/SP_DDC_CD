```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv("test.csv", skipinitialspace=True)
```

```python
df
```

```python
df['Salary'].fillna(df['Salary'].mean(), inplace=True)
```

```python
df['Experience'].fillna(df['Experience'].median(), inplace=True)
```

```python
df['Performance_Score.'].fillna(df['Performance_Score.'].mode()[0], inplace=True)
```

```python
df[df.duplicated()]
df.drop_duplicates(inplace=True)
```

```python
df['Joining_Date'] = pd.to_datetime(df['Joining_Date'],format='mixed')
```

```python
df
```

```python
sns.pairplot(df, hue="Department", diag_kind="kde")
plt.suptitle("Pairplot of Employee Data", y=1.02)
plt.xlabel("Features")
plt.ylabel("Values")
plt.show()
```

```python
# Select numeric columns
numeric_df = df.select_dtypes(include=['float64', 'int64'])

corr = numeric_df.corr()

plt.figure(figsize=(10, 6))

img = plt.imshow(corr)

plt.colorbar()

labels = corr.columns
plt.xticks(range(len(labels)), labels, rotation=45)
plt.yticks(range(len(labels)), labels)

# Add values inside boxes
for i in range(len(labels)):
    for j in range(len(labels)):
        plt.text(j, i, f"{corr.iloc[i, j]:.2f}",
                 ha='center', va='center')

plt.title("Feature Correlation Heatmap (Set 1)")
plt.tight_layout()
plt.show()
```

```python
import seaborn as sns
import matplotlib.pyplot as plt

df["Salary"] = pd.to_numeric(df["Salary"], errors='coerce')

plt.figure(figsize=(10, 6))

sns.boxplot(
    x="Department",
    y="Salary",
    data=df,
    palette="coolwarm"
)

# Add individual data points
sns.swarmplot(
    x="Department",
    y="Salary",
    data=df,
    color="black",
    size=4
)

plt.title("Salary Distribution with Data Points", fontsize=14, fontweight='bold')
plt.xlabel("Department")
plt.ylabel("Salary")

plt.xticks(rotation=30)
plt.tight_layout()
plt.show()
```

```python
import seaborn as sns
import matplotlib.pyplot as plt

plt.figure(figsize=(9,5))

sns.violinplot(
    y="Department",
    x="Salary",
    data=df,
    palette="pastel",
    inner=None   # remove inner lines for different look
)

plt.title("Salary Distribution (Horizontal View)")
plt.xlabel("Salary")
plt.ylabel("Department")

plt.tight_layout()
plt.show()
```
